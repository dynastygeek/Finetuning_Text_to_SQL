```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering

model_name = "distilbert-base-cased-distilled-squad"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForQuestionAnswering.from_pretrained(model_name)

context = """
Python is a programming language.
It was created by Guido van Rossum.
My name is prachi and I am a software engineer.
"""

# check if we can give a document and a question to the model and get an answer back

question = "What is my name?"

inputs = tokenizer(question, context, return_tensors="pt")

outputs = model(**inputs)

start = outputs.start_logits.argmax()
end = outputs.end_logits.argmax()

answer = tokenizer.decode(inputs["input_ids"][0][start:end+1])

print(answer)
```
